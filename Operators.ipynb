```python
#Check for Admission Criteria (Age>= 18 and have_degree=True)
age = 17
have_degree = True
 
#check for both conditions to be true
print(age >= 18 and have_degree == True)

#Check for one condition to be true
print(age >= 18 or have_degree == True)

#Check for 'not' condition
print (not age >= 18 and have_degree == True)
print (not age >= 18 or have_degree == True)
print (not age >= 18 and not have_degree == True)
print (not age >= 18 or not have_degree == True)
```

# For Loop
For <variable> in <iterable>


```python
names=['Antonia', 'Francis','stacey','Michelle','steve']
print(names)
for name in names:
    print(name)

word='python'
for letter in word:
    print(letter)

for name in names:
    for letter in name:
        if letter[0]=='A':
            print(name)
```

# Control flow
This refers to the order in which python statements are executed in a program
It allows a programto make decisions,repeat actions or skip sections of code depending on the condition instead of running the code top-down.

1. Conditional statements (Condition-Making)
    1. is statement
    2. if-else statement
2.  loops(repetition)
    1. For Loop(used when you know how many times to repeat/iterate)
    2. while loop( iterates/repeats as long as a condition is true)
3. Loop control Statements
1. break(stops loop completely)
2. continue (skip the current iteration and move to the next)
3. pass(does nothing;like a placeholder)

# Control flows help filter data based on conditions,apply transformations to datasets,automate repetitive tasks etc.

# While loop
# Keeps iterating until a certain condition becomes false

## Syntax
while condition:
>statements

1. The condition is checked before each iteration.<br>
2. If the condition is true the loop executes.
3. If condition is false the loop terminates.

```python
count=1
while count<=3:
    count+=1
    print(count,'Try login Again')
print('Your login trials have been exhausted.Try and contact the admin')

#Break;note that the while loop already has a break function this is just to show the use of break.

count=1
while count<=3: 
    if count==3:
        break
    print(count,'Try login Again')
    count+=1
```
