## Requirements Gathering and Analysis

### Functionalities of the system
1. Identify key Users and storing their data.
2. Identify the various Savings Goals for the different users.
3. Identify the set dates/due dates to achieve the goals.
4. Monitor the savings progress for the different users.
5. Send notifications to users regarding their status( both on going and final status)

### Variables
1. User ID
2. savings goal amount 
3. Achievement Date/ Due date
4. Amount saved
5. End Goal status(whether reached or not)

### Input and output requirements for sytem functionalities listed
1. The system is expected to identify and create different users.
2. The system should retrieve the users' saving goals and store the required data.
3. The system is required to create the achievement dates of the said goals.
4. The system is required to update the goal status for the respective users for the period before the date is reached.
5. After the achievement date is reached it is expected to come up with the summary of saving goal status of the individual users and come up with a conclusion of whether it was achieved or not.
6. The final status should clearly indicate whether the savings goal was achieved or not for different users and be able to compose and send notifications to the different users.




## Design

### Pseudocode
Function savingsGoal(User ID,savings goal amount,Achievement Date,Amount saved)
>Create a new savings goal status using the following data:User ID,savings goal amount,Achievement Date,Amount saved.
>store the record in the database.
>Return a unique identifier for the newly created savings goal.

Function savingsgoal progress(User ID,Amount saved)
>Retrieve the savings goal from the data base
>Update the total amount saved by adding cummulative savings up to date.
>calculate total amount saved in percentage based on expected savings amount and actual amount saved.
>store the records in the database.

Function endgoalsaving status(Amount saved)
>Retrieve data from database showing end date goal status.
>Calculate in percentage the actual amount saved versus the expected savings amount.
>if savings==100% then mark 'Achieved'
>else mark 'not achieved'

Function endsavingsgoal notification(User ID,Amount saved)
>Retrieve current savings goal status record
>Send current status notification to the respective users.
>if achievement date is closing up and savings goal is not achieved:
  >send a reminder to the user about their saving goals.
>else send a conglatulatory message on achievement.

Function endsavingsgoal notification(User ID,Amount saved)
>Retrieve final savings goal status record
>Send final status notification to the respective users.
>if savings == 'achieved'
  >send 'Congratulations on goal status achievement'
>else send 'goal status not achieved'

Terminate all

## Explanation
1. Creating a database containing user ID and goal status information.
2. Regular updating of the current goal status for the user using percentages.
3. Calculating the final status in percentage for the user and marking it as achieved or not achieved.
4. Creating reminder notifications for user if the goal has not yet been achieved and due date is near.
5. Creating a final end date notification to congratulate goal achievements by user else an unfortunate message for user.
6. Termination/compltion of the program.
