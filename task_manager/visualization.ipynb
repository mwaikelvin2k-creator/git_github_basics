# Visualization
Data representation in picture form
in python we use matplot lib

1. numeric data 
    discrete data - numeric data that is countable : bar charts, column charts.
    continuous data - data that is measurable : we use scatter plots or a histogram
2. Categorical data - 
    string
    numerical - represented by numbers eg: Trying to show the population of counties,5,001.

## Introduction to matplotlib
Its a library used to create graphs and other visualization.
Example : In  a retail store you want to visualize revenue againsts sales over time.

key components
1. plotting functions
2. axes
3. objects: bar, line, markers.
4. customation: fonts,colors,line styles,dotted.

```python
## creating a line chart
import matplotlib.pyplot as plt
import numpy as np

xpoints = np.array([0,10])
ypoints = np.array([0,250])

plt.plot(xpoints, ypoints)
plt.show()

```

```python
# customization
import matplotlib.pyplot as plt
import numpy as np

ypoints1 = np.array([3,8,2,7])
ypoints2 = np.array([1,4,7,2])
plt.plot(ypoints1,label = 'line1', linestyle = 'dotted', marker = 's')
plt.plot(ypoints2,label = 'line2', linestyle = 'dashed', color = 'red')
plt.title('DATA', fontweight = 'bold', color = 'red') #Title
plt.legend()
plt.ylabel('data y axis')
plt.xlabel('data x axis')
plt.show()
```

```python
# scatter plot
import matplotlib.pyplot as plt
import numpy as np

x = np.array([20,12,25,67,89,45]) # age
y = np.array([45,69,40,70,80,40]) # weight

plt.scatter(x, y)
plt.grid()
plt.show()
```

```python
# Bar plot
import matplotlib.pyplot as plt
import numpy as np

x = np.array(['Dennis','kelvin','Damaris','Majembe'])
y= np.array([60,75,65,80])

plt.bar(x,y)
plt.show()
```

```python
# Histogram
# shows frequencies of distribution
import matplotlib.pyplot as plt
import numpy as np

ages = [22, 25, 47, 35, 46, 52, 31, 28, 40, 38,
        55, 29, 43, 37, 50, 24, 33, 45, 27, 60]


plt.hist(ages, bins = 5, edgecolor = 'black')
plt.ylabel('Frequency')
plt.xlabel('age')
plt.title('DATA')
plt.show()
```

```python
# Piechart
import matplotlib.pyplot as plt
import numpy as np

y = np.array([43,24,56,80])
plt.pie(y, explode=[0.1,0,0,0], wedgeprops={'edgecolor':'black', 'width':0.5})
plt.legend('meme')
plt.show()


```
