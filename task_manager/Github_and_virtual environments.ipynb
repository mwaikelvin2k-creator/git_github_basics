# INTRODUCTION TO VIRTUAL ENVIRONMENTS
It is an isolated workspace for python projects. It lets you install python packages for one project without affecting other projects or system wide python.

project A == python 3.14
project b == pandas 2.7

Without a virtual environment the packages are bound to crash but with a virtual environment each project keeps its own packages.

With a virtual environment is like giving each python environment:
1. Its Own interpreter.
2. Package versions.
3. Its own libraries.



Before python packages python developers often installed packages (globally) using pip. This caused many problems where one project would break another, different projects needed different versions of the same package.
Suppose you install:
pip install django == 2.2
Then later on you install another version of django:
pip install django == 3.1
The old project is bound to stop working.
Virtual environments were created to solve this problem.



```python
There are various ways to create virtual environments:
1. Using python venv : which is a built in tool for creating virtual environments.
2. Using Anaconda : which is more powerful than venv.
```

## Using Venv
Step 1 : check version on gitbash (python --version)
step 2 : Create content folder command (mkdir)
Step 3 : Navigate to the new folder (cd)
step 4 : Create the virtual environment in the folder (python -m venv [venv_test])
Step 5 : Activating the environment
  * **Windows command**: source name_venv/scripts/activate
  * **Liux and mac**: source path_to_env/bin/activate
Step 6 : Installing dependancies.
