# Measures of central Tendancies
It summarizes a dataset by pinpointing a single value that represents its central location.


```python
import math
import numpy
# 1. Mean
#The average value of a certain dataset
#Example:

students = [40,64,67]
print((sum(students))/len(students))
    
#2. mode
def studentmode(nums):
    counts = {}
    for n in nums:
        if n in counts:
            counts[n]+=1
        else:
            counts[n] = 1
    max_count = max(counts.values(), default=0)
    modes = [n for n, count in counts.items() if count == max_count]
    return modes


#3. median
def mediancalculator(scores):
    scores1 = sorted(scores)
    lenght = len(scores1)
    if lenght%2 ==1:
        median = scores1[lenght//2]
    else:
        mid1 = scores1[(lenght//2)-1]
        mid2 = scores1[lenght//2]
        median = (mid1+mid2)/2
    return median


nums = [37,44,55,22,10,16,12,37]
#pep = mediancalculator(nums)
#print(pep)
```

```python
import statistics
students = [37,44,55,22,10,16,12,37]
# mode
mode=statistics.mode(students)
print(mode)

#median
median = statistics.median(students)
print(median)

#mean 
mean = statistics.mean(students)
print(mean)



```

```python
import numpy as np
from scipy import stats

students = [37,44,55,22,10,16,12,37]
# mode
mode = stats.mode(students)
print(mode)

#median
median = np.median(students)
print(median)

#mean 
mean = np.mean(students)
print(mean)
```

```python
## Measures of disoersion
#They include standard deviation, Variance, range, Interquatile range.

#standard_dev & Variance
petal =[5.0,6.0,5.5,7.0,4.5,3.2,8.5,3.5,5.0,3.1,4.7,8.0,5.6,5.3,5.1]
mean = sum(petal)/len(petal)
squared = [(x-mean)**2 for x in petal]
variance = sum(squared)/(len(petal)-1)
standard_deviation = variance**0.5
print(variance)
print(standard_deviation)

#InterQuatile range
import numpy as np
q1 = np.quantile(petal, 0.25)
q2 = np.quantile(petal, 0.50)
q3 = np.quantile(petal, 0.75)
interquatile_range = q3-q1

print(q1)
print(q2)
print(q3)
print(q3-q1)

```
