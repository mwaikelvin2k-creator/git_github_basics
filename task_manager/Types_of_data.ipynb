# INTRODUCTION TO TYPES OF DATA AND DATA COLLECTION

One very important skill in data science is understanding what kind of data you have. They can range from strings, integers, decimals, booleoans etc.
Choosing the right data type matters because it affects how you clean analyze and interpret data.
There are two broader categories:
1. Categorical(at times referred to as qualitative data)
2. Numerical(at times referred to as quantitative) 

## Categorical
Represents names,labels or groups
There are two types of categorical data:
1. Nominal data : Categorical data that has no natural order (one category is not higer than the other) eg.color,blood group.
Sometimes they may be stored as numbers but the numbers are stored as a codes representing the data.
2. Ordinal data : categorical data but the categories have a meaningful order but the difference between the categories cannot be measured. eg. education levels.
3. Binary categorical: Categorical data with only two types eg.yes/no,male/female

## Numerical
Data that is measurable. Actual quantities.
1. Discrete :Numerical data that comes from counting
2. Numerical :Numerical data that comes from measurement eg. weight, height, stock price etc.






There a number of methods you can use to clean data. [.split() .strip() .upper() .lower()]

## Tabular Data: The csv file
CSV(comma seperated values): Stores data as plain text where each record(row) and each column is an attribute.

name, age, city 
Francis, 34, Nairobi
Kelvin, 29, Ruiru
Kezzy, 19, Thika

CSV in built methods:
1. csv.reader() - It reads rows as a list
2. csv.DictReader() - Reads rows as dictionaries using the column names.
3. csv.writer() - Writes rows as csv
4. csv.DictWriter() - Writes dictionaries as csv




```python
import csv
with open('csvfile.csv', 'r' ) as file: # file is csvfile.csv
    css = csv.reader(file)
    for c in css:
        print(c)


with open('csvfile.csv', 'r' ) as file: 
    css = csv.DictReader(file)
    for c in css:
        print(c)


# with open('csvfile.csv', 'r' ) as file: 
    #css = csv.writer(file)
    #for c in css:
        #print(c)


#with open('csvfile.csv', 'r' ) as file: 
    #css = csv.DictWriter(file)
    #for c in css:
        #print(c)
```
