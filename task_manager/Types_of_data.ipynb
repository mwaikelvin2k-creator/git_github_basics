# INTRODUCTION TO TYPES OF DATA AND DATA COLLECTION

One very important skill in data science is understanding what kind of data you have. They can range from strings, integers, decimals, booleoans etc.
Choosing the right data type matters because it affects how you clean analyze and interpret data.
There are two broader categories:
1. Categorical(at times referred to as qualitative data)
2. Numerical(at times referred to as quantitative) 

## Categorical
Represents names,labels or groups
There are two types of categorical data:
1. Nominal data : Categorical data that has no natural order (one category is not higer than the other) eg.color,blood group.
Sometimes they may be stored as numbers but the numbers are stored as a codes representing the data.
2. Ordinal data : categorical data but the categories have a meaningful order but the difference between the categories cannot be measured. eg. education levels.
3. Binary categorical: Categorical data with only two types eg.yes/no,male/female

## Numerical
Data that is measurable. Actual quantities.
1. Discrete :Numerical data that comes from counting
2. Numerical :Numerical data that comes from measurement eg. weight, height, stock price etc.






There a number of methods you can use to clean data. [.split() .strip() .upper() .lower()]

## Tabular Data: The csv file
CSV(comma seperated values): Stores data as plain text where each record(row) and each column is an attribute.

name, age, city 
Francis, 34, Nairobi
Kelvin, 29, Ruiru
Kezzy, 19, Thika

CSV in built methods:
1. csv.reader() - It reads rows as a list
2. csv.DictReader() - Reads rows as dictionaries using the column names.
3. csv.writer() - Writes rows as csv
4. csv.DictWriter() - Writes dictionaries as csv




```python
import csv
with open('csvfile.csv', 'r' ) as file: # file is csvfile.csv
    css = csv.reader(file)
    for c in css:
        print(c)


with open('csvfile.csv', 'r' ) as file: 
    css = csv.DictReader(file)
    for c in css:
        print(c)


# with open('csvfile.csv', 'r' ) as file: 
    #css = csv.writer(file)
    #for c in css:
        #print(c)


#with open('csvfile.csv', 'r' ) as file: 
    #css = csv.DictWriter(file)
    #for c in css:
        #print(c)
```

## Data collection and Sampling Techniques
Data collection is the process of gathering dara about a population.
Sampling is the process of taking a subset(sample) from a population data.

population

The full group one is interested in for example: All students sitting KCSE exam in 2026 is 1000000 students.

Sample

A smaller group taken from the populatiom to be studied/ analyzed
Example: 1000 students selected from 48 schools across the country totalling to 40000

Sampling Bias

I t is when the population does not represent the population fairly which lead to misleading conclusions.
Example: You sample more gender than the other, sample more students from one region neglecting other regions.

Sampling matters in that it helps analyzing data. Good analysis should generalize the real world scenario which is only possible if the sample represents the whole population well.

The main sampling techniques are:
1. Simple Random sampling
Each member of the population has an equal chance of being selected.
Example: Randomly choose 100 students from a list of students.
It is normally applicable when the population is fairly known and you have access to the entire population (school list)

2. Stratified sampling
You divide population into stratas then you randomly sample each group.
Example: split customers by characteristics eg: agegroup, gender then sample each group.

It is used when you want the important groups/sections of your population represented and there are known groups in the population.

3. Cluster Sampling
You divide the population in clusterd often identified by features such as location/convinience then you randomly choose some clusters and sample everyone/most people in the chosen clusters.
Example: Choose a few schools froms everal provinces then survey/part of studentd in the schools

4. Sequential Sampling
You collect data step by step and decide whether to continue based on the results provided do far.
Example: Interviewing customers untill a certain outcome/pattern changes.
