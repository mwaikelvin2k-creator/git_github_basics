## Dictionaries
Data structure that stores data in key value pairs.
The key is the label used to access the value.

Dictionary syntax: dict_name = {key:value}



```python
# Example Dictionary
students = {
    'name':'Karen',
    'age':21,
    'gender':'female',
    'studentId':'m89',
    'course': 'Data science'
}
# we access data in a dictionary using the key which returns the value.
# Accessing name
print(students['name'])
print(students['course'])
print(students['name'],students['age'])

#using fstring
print(f'students name: {students['name']} students age: {students['age']}')
print('students name:',students['name'], 'students age:' ,students['age'])


#Get method
#Its safer to use the .get() method to avoid key error.
print(students.get('cohort'))
print(students.get('age'))

# methods associated with dictionaries
# .get(), .remove(), .replace(), .items(), .keys(), .values(), .update(), .copy()

##  .key(), .values(), .items()

print(students.keys()) #returns all keys in the dictionary
print(students.values())#returns all values in the dictionary
print(students.items())#returns all items in the dictionary

## .update()
students.update({'name': 'Francis'})
students.update({'age': 33})
print(students['name'])
print(students['age'])

## .pop()
students.pop('name')
print(students)

# remove & replace do not work in dictionaries

# clear()-remove everything
students.clear()
print(students)

#looping a Dictionary
for key, value in students.items():
    print(f'{key} {value}')

student_Details = [{
    'name':'Karen',
    'age':21,
    'gender':'female',
    'studentId':'m89',
    'course': 'Data science'
},
{
    'name':'john',
    'age':20,
    'gender':'male',
    'studentId':'m9',
    'course': 'Maths'
},
{
    'name':'Kelvin',
    'age':25,
    'gender':'male',
    'studentId':'m80',
    'course': 'Statistics'
}]
 

for stude in student_Details:
    for key,value in stude.items():
        print(f'{key}: {value}')


person = {'name': 'Emma', 'age': 28}
person.update({'country':'USA'})
print(person)

len("Hello, World!")


i = 5
while i > 0:
  print("countdown: ", i)
  if i == 3:
    print("Stopping countdown")
    break
  i-=1






```
