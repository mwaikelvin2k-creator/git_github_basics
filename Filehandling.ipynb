## File Handling and logging
They are essential for interacting with the computers file system and recording program activity.

file handling: refers to working with files on your computer such as reading, writing, modifying data stored in files.
logging: is the process of recording events errors, andmessages that occur during program execution.


```python
# opening a file
file = open('notes.txt', 'r')

# read the file
content = file.read()
print(content)
print(file.read())

# Writing the file
file = open('notes.txt','w')
file.write('This is a Data Science class')
file = open('notes.txt')
file.read()

file = open('notes.txt', 'w')
file.write('My name is kelvin')
file = open('notes.txt')
file.read()
print(file)

# file close
file.close()

# append
file = open('notes.txt','a')
file.write(' i am 24 yrs old.')
file = open('notes.txt')
file.readline()


file = open('notes.txt')
for line in file:
    print(line)

file.close()
```
